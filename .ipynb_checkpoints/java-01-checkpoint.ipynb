### Classe con Main e variabili locali

```java
public class Example {
    // var statiche primitive
    static float f1; // dichiarazione (inizializzazione implicita a 0.0f)
	static float f2 = 3.4f; // dichiarazione e inizializzazione
	// var statiche non primitive
    static String s1; // dichiarazione (inizializzazione implicita a null)
	// costanti statiche 
    static final double PI = 3.14; // dichiarazione e inizializzazione
	
    public static void main(String[] args) {
        // var locali primitive
		int i1 = 6; // dich. e iniz.
		int i2; // solo dich.
		// 3 metodi per inizializzare un char
        char c1 = 'V';
		char c2 = 86;
		char c3 = '\u0056';
        // var locali non primitive
		String s2;
			
		// stampa var loc 
        System.out.println("i1: " + i1);
        // stampa var statica globale
		System.out.println("f1: " + f1);
		System.out.println("f2: " + f2);
        // stampa concatenata (char inizializzati in 3 modi diversi)
		System.out.println("c1: " + c1 + " c2: " + c2 + " c3: " +c3);
        /* // stampa var loc non inizializzata        
         System.out.println("i2: " + i2); // compile-time error
        // stampa var loc non inizializzata
		System.out.println("s2: " + s2); // compile-time error
        // stampa var glob non inizializzata
		System.out.println(s1.toString()); // run-time error
        // assegnamento su una costante final
		PI = 3.0; // compile-time error
		*/
		int k = 0;
		System.out.println(true | k++ == 0);
		System.out.println("k (standard eval): " + k);
		k = 0;
		System.out.println(true || k++ == 0);
		System.out.println("k (short-circuit eval): " + k);
		//
		int n = 4;
		n ++;
		// int n = 8; // compile-time error
		{
			int m = 7;
			m ++;
		}
		{
			int m = 8;
			m --;
		}
		System.out.println("n: " + n);
		// System.out.println("m: " + m); // compile-time error
	}
}
Main.main(null);
```
